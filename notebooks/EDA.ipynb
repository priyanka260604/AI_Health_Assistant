```python
all_symptoms = pd.concat(
    [df[col] for col in symptom_columns]
)
```

```python
all_symptoms = all_symptoms.dropna()
```

```python
all_symptoms.nunique()
```

```python
all_symptoms.value_counts().head(20)
```

```python
top_symptoms = all_symptoms.value_counts().head(20)

plt.figure(figsize=(12,6))

plt.bar(top_symptoms.index, top_symptoms.values)

plt.xticks(rotation=90)

plt.title("Top 20 Symptoms")

plt.show()
```
